```python
import re
import pandas as pd

# -----------------------------
# Sentence Tokenizer
# -----------------------------
def split_sentences(text):
    """
    Split text into sentences using common sentence-ending punctuation.
    Works reasonably well for Indic languages.
    """
    sentences = re.split(r'(?<=[.!?।॥])\s+', text.strip())
    return [s.strip() for s in sentences if s.strip()]


# -----------------------------
# Build Dataset
# -----------------------------
MAX_SENTENCES = 1000

all_data = []

languages = list(dataset.keys())

print(f"Found {len(languages)} languages.\n")

for lang in languages:

    print(f"Processing {lang}...")

    sentence_count = 0

    for sample in dataset[lang]:

        # Get paragraph text
        paragraph = sample["text"]

        # Skip empty paragraphs
        if not paragraph or not paragraph.strip():
            continue

        # Split paragraph into sentences
        sentences = split_sentences(paragraph)

        for sentence in sentences:

            sentence = sentence.strip()

            # Skip very short sentences
            if len(sentence) < 5:
                continue

            all_data.append({
                "sentence": sentence,
                "label": lang
            })

            sentence_count += 1

            if sentence_count >= MAX_SENTENCES:
                break

        if sentence_count >= MAX_SENTENCES:
            break

    print(f"Collected {sentence_count} sentences.\n")


# -----------------------------
# Save CSV
# -----------------------------
df = pd.DataFrame(all_data)

df.to_csv("language_dataset.csv", index=False, encoding="utf-8")

print("=" * 50)
print("Dataset creation completed!")
print(f"Total samples : {len(df)}")
print(df["label"].value_counts())
print("=" * 50)
```

```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv("language_dataset.csv")

train_list = []
val_list = []
test_list = []

# For reproducibility
np.random.seed(42)

# Process each language separately
for label in df["label"].unique():

    # Select all samples of one language
    lang_df = df[df["label"] == label].copy()

    # Shuffle the samples
    lang_df = lang_df.sample(frac=1, random_state=42).reset_index(drop=True)

    # Split
    train = lang_df.iloc[:800]
    val = lang_df.iloc[800:900]
    test = lang_df.iloc[900:1000]

    train_list.append(train)
    val_list.append(val)
    test_list.append(test)

# Merge all languages
train_df = pd.concat(train_list, ignore_index=True)
val_df = pd.concat(val_list, ignore_index=True)
test_df = pd.concat(test_list, ignore_index=True)

# Shuffle each split (optional but recommended)
train_df = train_df.sample(frac=1, random_state=42).reset_index(drop=True)
val_df = val_df.sample(frac=1, random_state=42).reset_index(drop=True)
test_df = test_df.sample(frac=1, random_state=42).reset_index(drop=True)

# Save CSV files
train_df.to_csv("train.csv", index=False, encoding="utf-8")
val_df.to_csv("validation.csv", index=False, encoding="utf-8")
test_df.to_csv("test.csv", index=False, encoding="utf-8")

print("Train:", len(train_df))
print("Validation:", len(val_df))
print("Test:", len(test_df))
```

```python
import math
from collections import Counter, defaultdict

class TFIDFVectorizer:

    def __init__(self):
        self.vocabulary = {}
        self.idf = {}

    # -----------------------------------
    # Extract Features
    # -----------------------------------
    def get_features(self, sentence):

        words = sentence.split()

        features = []

        # Word Unigrams
        features.extend(words)

        # Word Bigrams
        for i in range(len(words) - 1):
            features.append(words[i] + "_" + words[i + 1])

        # Character 2,3,4 grams
        text = sentence.replace(" ", "")

        for n in [2, 3, 4]:
            for i in range(len(text) - n + 1):
                features.append(text[i:i+n])

        return features

    # -----------------------------------
    # Build Vocabulary + IDF
    # -----------------------------------
    def fit(self, sentences):

        document_frequency = defaultdict(int)

        total_documents = len(sentences)

        for sentence in sentences:

            features = self.get_features(sentence)

            unique_features = set(features)

            for feature in unique_features:
                document_frequency[feature] += 1

        # Vocabulary
        for index, feature in enumerate(document_frequency):
            self.vocabulary[feature] = index

        # IDF
        for feature, df in document_frequency.items():
            self.idf[feature] = math.log(total_documents / (df + 1)) + 1

    # -----------------------------------
    # Sparse Transform
    # -----------------------------------
    def transform(self, sentences):

        vectors = []

        for sentence in sentences:

            features = self.get_features(sentence)

            tf = Counter(features)

            total_terms = len(features)

            # Sparse dictionary instead of dense NumPy vector
            vector = {}

            for feature, count in tf.items():

                if feature not in self.vocabulary:
                    continue

                index = self.vocabulary[feature]

                tf_value = count / total_terms

                vector[index] = tf_value * self.idf[feature]

            vectors.append(vector)

        return vectors

    # -----------------------------------
    # Fit + Transform
    # -----------------------------------
    def fit_transform(self, sentences):

        self.fit(sentences)

        return self.transform(sentences)
```

```python
import pandas as pd

train = pd.read_csv("train.csv")
validation = pd.read_csv("validation.csv")
test = pd.read_csv("test.csv")

vectorizer = TFIDFVectorizer()

X_train = vectorizer.fit_transform(train["sentence"].tolist())

X_val = vectorizer.transform(validation["sentence"].tolist())

X_test = vectorizer.transform(test["sentence"].tolist())

print("Vocabulary Size :", len(vectorizer.vocabulary))
print("Training Samples:", len(X_train))
print("Validation Samples:", len(X_val))
print("Test Samples:", len(X_test))

print("\nFirst Sparse Vector:")
print(X_train[0])
```

```python
import numpy as np

# Get unique language labels
classes = sorted(train["label"].unique())

# Label -> Integer
label_to_index = {label: i for i, label in enumerate(classes)}

# Integer -> Label
index_to_label = {i: label for label, i in label_to_index.items()}

# Encode labels
y_train = np.array([label_to_index[x] for x in train["label"]])
y_val = np.array([label_to_index[x] for x in validation["label"]])
y_test = np.array([label_to_index[x] for x in test["label"]])

print("Classes:", len(classes))
print(classes)
```

```python
import numpy as np

class LogisticRegressionScratch:

    def __init__(self,
                 num_features,
                 num_classes,
                 learning_rate=0.5,
                 epochs=15):

        self.num_features = num_features
        self.num_classes = num_classes
        self.learning_rate = learning_rate
        self.epochs = epochs

        # Weight matrix
        self.W = np.zeros((num_classes, num_features))

        # Bias
        self.b = np.zeros(num_classes)

    # ------------------------
    # Softmax
    # ------------------------

    def softmax(self, scores):

        scores = scores - np.max(scores)

        exp_scores = np.exp(scores)

        return exp_scores / np.sum(exp_scores)

    # ------------------------
    # Training
    # ------------------------

    def fit(self, X, y):

        for epoch in range(self.epochs):

            correct = 0

            for vector, target in zip(X, y):

                scores = np.copy(self.b)

                # Sparse dot product
                for feature, value in vector.items():
                    scores += self.W[:, feature] * value

                probs = self.softmax(scores)

                prediction = np.argmax(probs)

                if prediction == target:
                    correct += 1

                # Gradient update
                probs[target] -= 1

                for feature, value in vector.items():
                    self.W[:, feature] -= self.learning_rate * probs * value

                self.b -= self.learning_rate * probs

            accuracy = correct / len(y)

            print(f"Epoch {epoch+1}/{self.epochs}  Accuracy = {accuracy:.4f}")

    # ------------------------
    # Prediction
    # ------------------------

    def predict(self, X):

        predictions = []

        for vector in X:

            scores = np.copy(self.b)

            for feature, value in vector.items():
                scores += self.W[:, feature] * value

            predictions.append(np.argmax(scores))

        return np.array(predictions)
```

```python
num_features = len(vectorizer.vocabulary)
num_classes = len(classes)

model = LogisticRegressionScratch(
    num_features=num_features,
    num_classes=num_classes,
    learning_rate=0.5,
    epochs=15
)

model.fit(X_train, y_train)
```

```python
def macro_f1_score(y_true, y_pred, num_classes):

    f1_scores = []

    for cls in range(num_classes):

        tp = 0
        fp = 0
        fn = 0

        for yt, yp in zip(y_true, y_pred):

            if yt == cls and yp == cls:
                tp += 1

            elif yt != cls and yp == cls:
                fp += 1

            elif yt == cls and yp != cls:
                fn += 1

        # Precision
        if tp + fp == 0:
            precision = 0
        else:
            precision = tp / (tp + fp)

        # Recall
        if tp + fn == 0:
            recall = 0
        else:
            recall = tp / (tp + fn)

        # F1
        if precision + recall == 0:
            f1 = 0
        else:
            f1 = 2 * precision * recall / (precision + recall)

        f1_scores.append(f1)

    return sum(f1_scores) / len(f1_scores)
```

```python
val_macro_f1 = macro_f1_score(
    y_val,
    val_predictions,
    len(classes)
)

print("Validation Macro-F1:", round(val_macro_f1, 4))
```
